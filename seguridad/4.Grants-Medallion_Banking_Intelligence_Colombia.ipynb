**Alcance:** `SELECT ON SCHEMA` concede lectura a los objetos aplicables del esquema, incluidos objetos futuros según la herencia de privilegios de Unity Catalog. Para un profesor con acceso limitado, sustituirlo por `GRANT SELECT ON TABLE` de objetos concretos. `SELECT` en una vista no necesariamente da acceso directo a sus tablas subyacentes.
